```python
array=np.array([1,2,3,4,5])
print(array)
print("type of array:",type(array))
print("data type:",array.dtype)
```

```python
array=np.array(array,dtype=np.int8)
print(array)
print("no of bytes:",array.nbytes)
```

"In int8 we can store value upto (-128 to 127)"

```python
float_demo=np.array([1.2,2.3,3.4,4.5])
print(float_demo)
print(type(float_demo))
print("no of bytes:",float_demo.nbytes)
```

```python
#string
array=np.array(["dinesh","yogitha","karthick","samar","suganth"])
print(array)
print("type of array:",array.dtype)
print("no of bytes:",array.nbytes)
```

```python
# 0-dimension
array=np.array(1)
print(array)
print("dimesion:",array.ndim)
```

```python
#1dimension

arr=np.array([1,2,3,4,5])


# 2-dimension

brr=np.array([
    [1,2,3],
    [4,5,6]
])


# 3-dimension

crr=np.array([
    [
        [1,2,3],
        [4,5,6]
    ],
    [
        [11,22,33],
        [12,13,14]
    ]
])


print(arr)
print("dimension:",arr.ndim)

print(brr)
print("dimension:",brr.ndim)

print(crr)
print("dimension:",crr.ndim)
```

```python
#to access the elements
#(layer,start,stop)
print(crr[1][1][1])
#or
print(crr[1,1,1])
```
