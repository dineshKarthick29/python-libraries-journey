```python
#aggregate function

marks=np.array([60,76,45,90,88])

print(np.sum(marks))
print(np.mean(marks))
print(np.median(marks))
print(np.min(marks))
print(np.max(marks))
print(np.std(marks))
print(np.var(marks))
print(np.argmin(marks))
print(np.argmax(marks))
```

```python
#filtering
import numpy as np
marks=np.array([50,99,65,45,76])
print(marks)
print(marks[marks>50])
print(marks>50)
pass_marks=marks[marks>50]
print(pass_marks)
print(marks[marks%2==0])

#multing filtering

even=marks[(marks>50) & (marks>70)]
print(even)

con=marks[(marks>=50) | (marks%2==0)]
print(con)



```

```python
#multiple filtering

arr=np.array([
    [1,2,3],
    [4,5,6]
])
even=arr[(arr%2==0)]
print(even)
```

```python
arr=np.where(arr%2==0,arr,0)
print(arr)
```
