```python
import pandas as pd
print(pd.__version__)
```

```python
#series - 1 dimensions
a=["goms","code","siva"]
series=pd.Series(a)
print(series)
```

```python
b=[19,20,17,15]
bs=pd.Series(b)
print(b)
print(bs)
print()
print(bs[bs>18])
```

```python
print(bs.loc[2])
```

```python
b=pd.Series(bs.values,index=['a','b','c','d'])
print(bs)
print(b)
```

```python
#we can also store as in series
c={
    "mokesh":2.5,
    "kaviyarashu":2.2,
    "sabarish":5
}
c_series=pd.Series(c)
print(c)
```

```python
#filtering in series
print(c_series[c_series>=2.5])
```

```python
#data frames
students={
    "name":["samar","sugan","karthi"],
    "age":[19,18,19],
    "12thmark":[369,335,389]
}
df=pd.DataFrame(students)
print(df)
```

```python
df=pd.DataFrame(students,index=['std 1','std 2','std 3'])
print(df)
print()
print(df.loc['std 2'])
print()
print(df.iloc[1])
```

```python
#to add new column
df["medium"]=["cbse","cbse","cbse"]
print(df)
```
